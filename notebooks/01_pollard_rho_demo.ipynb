## Pollard's Rho Factorisation

Pollard's rho (1975) finds a non-trivial factor of *n* in expected **O(n^{1/4})**
time using Floyd's cycle detection on the sequence x_{k+1} = (x_k² + c) mod n.

| RSA modulus bits | Smallest prime factor | Expected Pollard steps | Feasibility |
|---|---|---|---|
| 512  | ~256-bit | ~2^64   | Hours on modern hardware |
| 1024 | ~512-bit | ~2^128  | Infeasible |
| 2048 | ~1024-bit| ~2^256  | Astronomically infeasible |

**Reference:** Pollard, J. M. (1975). *A Monte Carlo method for factorization.*
BIT Numerical Mathematics, 15(3), 331–334.

```
import time
from rsa_attacks.pollard_rho import factorize

# 2^64 + 1 = F6 (6th Fermat number) = 274177 × 67280421310721
# Smallest factor is 18-bit  →  O(2^4.5) ≈ 512 expected steps
F6 = 18_446_744_073_709_551_617

start   = time.perf_counter()
factors = factorize(F6)
elapsed = time.perf_counter() - start

print(f"n        = {F6}")
print(f"factors  = {factors}")
print(f"verified = {factors[0] * factors[1] == F6}")
print(f"time     = {elapsed:.4f}s")
```

```
import sympy

# Two consecutive 64-bit primes  →  128-bit semiprime
# Smallest factor ≈ 2^63  →  O(2^(63/4)) ≈ 2^15.75 ≈ 57 000 expected steps
p = sympy.nextprime(2**63)
q = sympy.nextprime(p)
n = p * q

print(f"p = {p}")
print(f"q = {q}")
print(f"n = {n}  ({n.bit_length()} bits)\n")

start   = time.perf_counter()
factors = factorize(n)
elapsed = time.perf_counter() - start

print(f"factors  = {factors}")
print(f"verified = {factors[0] * factors[1] == n}")
print(f"time     = {elapsed:.4f}s")
```

## Why RSA Needs 2048+ Bits

For a 2048-bit RSA modulus, the prime factors are each ~1024 bits (≈ 10^308).
Pollard's rho would need ~2^256 steps — more than the number of atoms in the
observable universe.

The best known algorithm for large RSA moduli is the **General Number Field Sieve
(GNFS)**, which runs in roughly L(n, 1/3) time (sub-exponential but super-polynomial).
NIST estimates 2048-bit RSA at ~112 bits of security; 3072 bits at ~128 bits.

The 64-bit and 128-bit semiprimes above fell in milliseconds or seconds because
their prime factors were small. Production RSA keys are designed so that the
smallest prime factor is ~half the modulus size — well beyond any feasible attack.
